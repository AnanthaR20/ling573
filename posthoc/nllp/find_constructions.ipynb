```python
from diversity import compression_ratio, homogenization_score, ngram_diversity_score
import pandas as pd
import nltk
# nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import time
```

```python
eng_stopwords = stopwords.words('english') 
```

```python
train = pd.read_csv("../../preprocess/data/billsum_train.csv")
```

```python
%%time
train['text'] = train['text'].apply(word_tokenize)
```

```python
%%time 
train['text'] = train['text'].apply(lambda words: [word for word in words if word not in eng_stopwords])
```

```python
%%time
train['text'] = train['text'].str.join(' ')
```
